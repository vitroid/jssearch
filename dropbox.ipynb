```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import time
from datetime import datetime

# deadline = time.time()
# deadline= datetime.strptime("2024/07/06 00:00", "%Y/%m/%d %H:%M").timestamp()
deadline = datetime.strptime("2024/07/06 00:00+0900", "%Y/%m/%d %H:%M%z").timestamp()
# datetime.strptime("2018/11/01 09:00+0900", "%Y/%m/%d %H:%M%z")

# print(deadline)
mtimes = np.array(
    [
        os.path.getmtime(file)
        for file in glob.glob("/Users/matto/Dropbox/Apps/pdfpreview/*.pdf")
    ]
)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

ax = axes[0]
ra = (-3, 27)
ax.hist((deadline - mtimes) / 3600 / 24, bins=24 * 30, range=ra)
ax.set_xlim(*ra)
ax.set_xlabel("Days before the deadline")
ax.set_ylabel("PDF previews / hour")

ax = axes[1]
now = time.time()
ra = (0, 24 + 48)
ax.hist((mtimes - deadline) / 3600 + 24, bins=24 * 6 * 3, range=ra)
ax.set_xlim(*ra)
ax.set_xlabel("OClock of the last day")
ax.set_ylabel("PDF previews / 10 mins")


# plt.xscale("log")
# plt.yscale("log")
```
